## d) generate weights and biases

```python
"""#### Layer 1

z1 = W1 @ x + b1

[20,100] = [20,200]@[200,100] + [20,1]

### Layer 2

z2 = W2 @ a1 + b2

[10,100] = [10,20]@[20,100] + [10,1]"""
W1 = np.random.randn(20,200)
b1 = np.random.randn(20,1)
W2 = np.random.randn(10,20)
b2 = np.random.randn(10,1)
W1.shape, b1.shape, W2.shape, b2.shape
```

## c) implement nn in python

```python
def forward(x):
    z1 = W1@x + b1
    a1 = np.tanh(z1)
    z2 = W2@a1 + b2
    return z2
```

> z2 = [10,100]

> labels = [100,1]

> X = [100,200], we have to transpose it

```python
X.shape
```

```python
X = X.T
X.shape
```

```python
output = forward(X)
output.shape
```

```python
y_pred = output.T
y_pred.shape # out task is to apply softmax on this
```

```python
def softmax(x):
    x_shifted = x - np.max(x, axis=1, keepdims=True)
    exp_x = np.exp(x_shifted)
    return exp_x / np.sum(exp_x, axis=1, keepdims=True)

# Example
y_out = softmax(y_pred)
y_out.shape
```

```python
row_sums = np.sum(y_out, axis=1)
print(row_sums)
```

## e) calculate loss

```python
pred_classes = np.argmax(y_out, axis=1)
pred_classes
```

```python
pred_classes_one_hot = np.zeros((100,10))
pred_classes_one_hot[np.arange(100), pred_classes] = 1
pred_classes_one_hot
```

```python
labels
```

```python
out = pred_classes_one_hot
out
```

```python
probs = out[np.arange(100), labels]
loss = -np.mean(np.log(probs + 1e-9))
loss
```

## f) calculate accuracy, precision, recall, f1 score

```python
pred_classes, labels
```

```python
accuracy = np.mean(pred_classes == labels)
accuracy
```

```python
classes = np.unique(labels)
classes
```

```python
precision_list = []
recall_list = []
f1_list = []

for c in classes:
    tp = np.sum((pred_classes == c) & (labels == c))
    fp = np.sum((pred_classes == c) & (labels != c))
    fn = np.sum((pred_classes != c) & (labels == c))
    
    precision = tp / (tp + fp + 1e-9)
    recall = tp / (tp + fn + 1e-9)
    f1 = 2 * precision * recall / (precision + recall + 1e-9)
    
    precision_list.append(precision)
    recall_list.append(recall)
    f1_list.append(f1)
```

```python
precision_list
```

```python
precision = np.mean(precision_list)
recall = np.mean(recall_list)
f1_score = np.mean(f1_list)
```

```python
print("Precision score ", precision)
print("recall score : ", recall)
print("f1 score ", f1_score)
```
